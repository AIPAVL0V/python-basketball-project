```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
from pathlib import Path
```

```python
data_raw_path = "../data/raw/nba.sqlite"
data_processed_path = Path("../data/processed")
```

```python
def clean_sql_table(table_name):
    data_processed_path.mkdir(parents=True, exist_ok=True)

    conn = sqlite3.connect(data_raw_path)

    df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
    
    conn.close()

    print(f"Shape before cleaning: {df.shape}")

    df_clean = df.copy()

    df_clean.columns = (
        df_clean.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
    )

    df_clean = df_clean.drop_duplicates()

    df_clean = df_clean.dropna(how="all")

    numeric_cols = df_clean.select_dtypes(include="number").columns
    text_cols = df_clean.select_dtypes(include="object").columns

    df_clean[numeric_cols] = df_clean[numeric_cols].fillna(
        df_clean[numeric_cols].median()
    )

    df_clean[text_cols] = df_clean[text_cols].fillna("unknown")

    output_path = data_processed_path / f"{table_name}.csv"

    df_clean.to_csv(output_path, index=False)

    print(f"Saved: {output_path}")

    print(f"Shape after cleaning: {df_clean.shape}")

    return df_clean
```

```python
conn = sqlite3.connect(data_raw_path)

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

for name in tables['name']:
  print(name)
  if (name == 'play_by_play'):
    continue
  clean_sql_table(name)
```

```python
clean_sql_table('play_by_play')
```
